```python
print("\n--- DATA TYPES ---")
print(df.dtypes)

print("\n--- COLUMNS ---")
print(df.columns.tolist())

print("\n--- SHAPE ---")
print(df.shape)

print("\n--- BASIC INFO ---")
df.info()
```

```python
print("\n--- MISSING VALUES ---")

missing = df.isnull().sum()

print(missing[missing > 0])
```

```python
print("\n--- DUPLICATE ROWS ---")

print("Duplicate rows:", df.duplicated().sum())
print("\n--- DUPLICATE ORDER IDs ---")

print(df["Order_ID"].duplicated().sum())
```

```python
print("\n--- CATEGORIES ---")

print(df["Category"].value_counts(dropna=False))
print(df["Payment_Mode"].value_counts(dropna=False))
print(df["Region"].value_counts(dropna=False))
```

```python
print("\n--- DATE RANGE ---")

print(df["Order_Date"].min())
print(df["Order_Date"].max())

print("\n--- DATE TYPE ---")

print(df["Order_Date"].dtype)
```

```python
print("\n--- INVALID QUANTITIES ---")

print(df[df["Quantity"] <= 0].head())

print("\n--- INVALID DISCOUNTS ---")

print(df[(df["Discount"] < 0) | (df["Discount"] > 1)].head())

print("\n--- INVALID SALES ---")

print(df[df["Sales"] < 0].head())
```

```python
print("\n--- EXTREME PROFIT VALUES ---")

print(df["Profit"].sort_values().head(10))
print(df["Profit"].sort_values(ascending=False).head(10))
```
